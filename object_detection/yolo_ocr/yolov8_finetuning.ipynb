```python
from ultralytics import YOLO
import shutil, os
```

```python
## loading the model
model = YOLO('yolov8n.pt')

## train
results = model.train(
    data=dataset.location + "/data.yaml",
    epochs = 2,
    imgsz=640,
    batch=16,
    workers = 2,
    device=0
)
```
